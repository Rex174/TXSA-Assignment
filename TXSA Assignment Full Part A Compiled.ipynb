## 1.1 Demonstrate word tokenisation using the split function, Regular Expression and NLTK packages separately and report the output. (2 marks)

```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
import string

# Load data from Data_1.txt
with open('..\Part_A\data\Data_1.txt', 'r') as file:
    text_data = file.read()

# 1. Tokenization using split() function
tokens_split = text_data.split()

# 2. Tokenization using Regular Expression (finds all alphanumeric sequences)
tokens_re = re.findall(r'\w+', text_data)

# 3. Tokenization using NLTK
# Ensure nltk resources are downloaded: nltk.download('punkt')
tokens_nltk = word_tokenize(text_data)

print("Split() Tokens:", tokens_split[:10])
print("Regex Tokens:", tokens_re[:10])
print("NLTK Tokens:", tokens_nltk[:10])
```

## 1.2 Justify the most suitable tokenisation operation for text analytics. Support your answer using obtained outputs. (3 marks)

Justification of the Most Suitable Tokenization Operation
NLTK tokenization is the most suitable for text analytics.
•	While split() is fast, it fails to separate punctuation from words (e.g., "input." remains as one token). Regular expressions are flexible but often require complex patterns to handle edge cases like contractions or abbreviations. NLTK’s word_tokenize is pre-trained to handle these complexities, such as correctly identifying sentence-ending punctuation as separate tokens while keeping decimal points or abbreviations intact.

## 1.3 Demonstrate stop words and punctuations removal and report the output suitably along with the stop words found in the given text corpus. (2 marks)

```python
# 2. Filtering
stop_words = set(stopwords.words('english'))
punctuations = set(string.punctuation)

# List only the unique stop words actually present in the corpus
found_stop_words = sorted(list(set([w.lower() for w in tokens_nltk if w.lower() in stop_words])))

# Filtered tokens (using NLTK as it is the most suitable)
filtered_tokens = [w for w in tokens_nltk if w.lower() not in stop_words and w not in punctuations]

print("\nUnique Stop Words Found:", found_stop_words)
print("\nFiltered Tokens (First 20):", filtered_tokens[:20])
```

## 1.4 Explain the importance of filtering the stop words and punctuations in text analytics. (3 marks)

Filtering is critical because: 
•	Noise Reduction: Stop words (e.g., "the", "is", "at") occur frequently but carry little semantic value, potentially drowning out meaningful keywords in tasks like topic modeling.
•	Dimensionality Reduction: Removing these tokens reduces the size of the dataset, which improves the computational efficiency of machine learning models.
•	Accuracy: Punctuation can often be misinterpreted as part of a word, leading to multiple variations of the same token (e.g., "tasks" vs "tasks.") which confuses the analysis.

# 2.1 Explain the importance of stemming in text analytics (3 marks)

The importance of stemming in text analytics is to reduce dimensionality and variety in text analytics. This is because stemming can successfully merge similar word forms such as ‘running’ and ‘ran’ from the base word ‘run’ hence reducing the vocabulary size and sparsity. Besides that, the next importance of stemming is to improve the recall ability for information retrieval. This is because stemming can help match different forms of the same lexical item hence a query for ‘run’ can match with ‘running’ and other variations. Moreover, stemming is also necessary to boost the accuracy and precision of NLP models as the resulting smaller and normalized vocabulary from stemming can make the models perform faster and produce less noise. Furthermore, reduction of dataset complexity from stemming can save costs from processing power and storage requirements necessary for the models.

# 2.2) Demonstrate word stemming using Regular Expression, Porter Stemmer and Lancaster Stemmer and report the output (3 marks)

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer, LancasterStemmer, RegexpStemmer


# 1. Load Data_1.txt
with open("..\Part_A\data\Data_1.txt", "r", encoding="utf-8") as f:
    data_1 = f.read()

# 2. Initialize stemmers
porter = PorterStemmer()
lancaster = LancasterStemmer()

# Implementing rule set for Regular Expression Stemmer
regex = RegexpStemmer('ing$|s$|ed$|able$|ion$|ment$', min=4)

# 3. Tokenize the Data_1.txt sentences into each word to undergo stemming
words = [w for w in word_tokenize(data_1)]

# Remove redundant words from output
words = sorted(list(set(words)))

# 4. Comparison table
print(" \n 2.2 Demonstrate word stemming using Regular Expression, Porter Stemmer and Lancaster Stemmer and report the output. \n")
print(f"{'Original':<20} | {'Regular Expression':<20} | {'Porter':<20} | {'Lancaster':<20}")
print("-" * 90)
for word in words:
  # Regular Expression Stemming
    r_stem = regex.stem(word)
  # Porter Stemming
    p_stem = porter.stem(word)
  # Lancaster Stemming
    l_stem = lancaster.stem(word)
  # Print table
    print(f"{word:<20} | {r_stem:<20} | {p_stem:<20} | {l_stem:<20}")
```

# 2.3) Explain the differences among Regular Expression stemmer, Porter Stemmer and Lancaster Stemmer using the obtained output (4 marks)

The first difference among Regular Expression Stemmer, Porter Stemmer and Lancaster Stemmer is their characteristic where Regular Expression Stemmer is a manual and rule-based stemming approach which only removes suffixes that had been specified in the code such as -ing, -s, -ed, -able, -ion and -ment. On the other hand, Porter Stemmer is the most gentle stemming algorithm most popular for most NLP projects and primarily accepted as the industry standard which uses more than 50 rules, grouped in 5 steps and some sub-steps. Lancaster stemmer is known as the most destructive stemming algorithm has more than 100 rules double of the Porter stemmer.

The next difference between them is that Regular Expression Stemmer operates in a fast but rigid manner as the words like `considered` had been stemmed correctly to `consider` by removing the `-ed` however words like `class` had been misspelled into `clas` due to the suffix `-s` removal rule. Porter Stemmer is known to be moderately aggressive that can preserve the original word’s recognizability such as stemming the word `example` to `exampl` and the word `sequence` to `sequenc`. However, Lancaster Stemmer is the most aggressive which creates the shortest but most distorted stems such as converting `variants` to `vary` changing the core word, `advance` to `adv` and `isolation` to `isol` causing extreme shortening.

The third difference is that Regular Expression stemmer is highly flexible and has customizable rules such as stemming `defined` into `defin` making it suitable to keep or remove particular suffixes for specific domain needs. Porter Stemmer provides the perfect balance between word reduction and readability hence being perfect for information retrieval as inflectional forms are returned to a common root like `inputs` and `input` being stemmed into `input`. Lancaster Stemmer is highly efficient for small datasets and quick preprocessing. 

The fourth difference between them is that Regular Expression stemmer requires significant manual tuning via proper rule definition and suffers from a high risk of under-stemming like `classification` to `classificat`. It is also computationally expensive for large datasets. Next, Porter Stemmer suffers from a high risk of creating non-dictionary words like `advnac` from `advance` and is limited to English words only. Lancaster stemmer suffers from a high risk of over-stemming with a high risk of changing the root word or meaning significantly like `number` being stemmed to `numb` and creating non-word stems like `sequ` from `sequence`.

## 3.1 DEMONSTRATE POS TAGGING (3 marks)

```python
# Import Libraries and Setup
import nltk
from textblob import TextBlob

nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
```

```python
# Load Input Sentence
with open("../Part_A/data/Data_2.txt", "r", encoding="utf-8") as f:
    sentence = f.read().strip()

print("Input Sentence:")
print(sentence)
```

### 3.1.1 NLTK POS Tagger

```python
print("\n3.1.1 NLTK POS TAGGER\n")

tokens = nltk.word_tokenize(sentence)
nltk_pos = nltk.pos_tag(tokens)

print("Tokenized Words:")
print(tokens)

print("\n\nNLTK POS Tagging Output:")
print(f"{'Word':<15} {'POS Tag':<10} {'Description'}")

pos_meanings = {
    'DT': 'Determiner',
    'JJ': 'Adjective',
    'NN': 'Noun, singular',
    'VBD': 'Verb, past tense',
    'IN': 'Preposition',
    'CC': 'Coordinating conjunction',
    'RB': 'Adverb',
    '.': 'Punctuation'
}

for word, tag in nltk_pos:
    meaning = pos_meanings.get(tag, 'Unknown')
    print(f"{word:<15} {tag:<10} {meaning}")

print("\n\nComplete Output:")
print(nltk_pos)
```

### 3.1.2 TextBlob POS Tagger

```python
print("\n\n3.1.2 TEXTBLOB POS TAGGER\n")

blob = TextBlob(sentence)
textblob_pos = blob.tags

print("TextBlob POS Tagging Output:")
print(f"{'Word':<15} {'POS Tag':<10} {'Description'}")

for word, tag in textblob_pos:
    meaning = pos_meanings.get(tag, 'Unknown')
    print(f"{word:<15} {tag:<10} {meaning}")

print("\n\nComplete Output:")
print(list(textblob_pos))
```

### 3.1.3 Regular Expression Tagger

```python
print("\n3.1.3 REGULAR EXPRESSION TAGGER\n")

patterns = [
    (r'\bThe\b|\bthe\b|\ba\b|\ban\b', 'DT'),
    (r'.*ing$', 'VBG'),
    (r'.*ed$', 'VBD'),
    (r'.*ly$', 'RB'),
    (r'\b(at|in|on|with|by|for|to)\b', 'IN'),
    (r'\b(and|or|but)\b', 'CC'),
    (r'\b(away|very|quite|rather)\b', 'RB'),
    (r'\.|\,|\!|\?', '.'),
    (r'.*', 'NN')
]

regexp_tagger = nltk.RegexpTagger(patterns)
regex_pos = regexp_tagger.tag(tokens)

print("Regular Expression POS Tagging Output:")
print(f"{'Word':<15} {'POS Tag':<10} {'Description'}")

for word, tag in regex_pos:
    meaning = pos_meanings.get(tag, 'Unknown')
    print(f"{word:<15} {tag:<10} {meaning}")

print("\n\nComplete Output:")
print(regex_pos)
```

```python
# Import Libraries for Parse Trees

from nltk import RegexpParser
from nltk.tree import Tree
```

```python
# Prepare Sentence and POS Tags
print("\nSentence:", sentence)

print("\nPOS Tags:")
for word, tag in nltk_pos:
    print(f"  {word:12} -> {tag}")
```

```python
# METHOD 1 - Chunk Parser
print("\n\nMETHOD 1: CHUNK PARSER\n")

chunk_grammar = r"""
    NP: {<DT>?<JJ>*<NN.*>+}
    VP: {<VB.*><RB>?}
    VP: {<VB.*><IN><NP>}
    PP: {<IN><NP>}
"""

chunk_parser = RegexpParser(chunk_grammar)
chunked_tree = chunk_parser.parse(nltk_pos)

print("Tree Output:")
print(chunked_tree)

print("\n\nPretty Print:")
chunked_tree.pretty_print()
```

```python
# METHOD 2 - Context-Free Grammar Parser
print("\n\nMETHOD 2: CONTEXT-FREE GRAMMAR PARSER\n")

cfg_grammar = nltk.CFG.fromstring("""
    S -> NP VP
    S -> NP VP CC VP
    
    NP -> DT ADJP NN
    NP -> DT NN
    
    ADJP -> JJ JJ
    ADJP -> JJ
    
    VP -> VBD PP
    VP -> VBD RB
    
    PP -> IN NP
    
    DT -> 'The' | 'the'
    JJ -> 'big' | 'black' | 'white'
    NN -> 'dog' | 'cat'
    VBD -> 'barked' | 'chased'
    IN -> 'at'
    CC -> 'and'
    RB -> 'away'
""")

parse_tokens = [word for word in tokens if word != '.']
parser = nltk.ChartParser(cfg_grammar)
parse_trees = list(parser.parse(parse_tokens))

print(f"Number of parse trees: {len(parse_trees)}")

for i, tree in enumerate(parse_trees, 1):
    print(f"\n\nParse Tree #{i}")
    print(tree)
    print("\nPretty Print:")
    tree.pretty_print()

```

```python
# METHOD 3 - Manual Parse Tree
print("\nMETHOD 3: MANUAL PARSE TREE\n")

manual_tree = Tree('S', [
    Tree('NP', [
        Tree('DT', ['The']),
        Tree('ADJP', [
            Tree('JJ', ['big']),
            Tree('JJ', ['black'])
        ]),
        Tree('NN', ['dog'])
    ]),
    Tree('VP', [
        Tree('VBD', ['barked']),
        Tree('PP', [
            Tree('IN', ['at']),
            Tree('NP', [
                Tree('DT', ['the']),
                Tree('JJ', ['white']),
                Tree('NN', ['cat'])
            ])
        ])
    ]),
    Tree('CC', ['and']),
    Tree('VP', [
        Tree('VBD', ['chased']),
        Tree('ADVP', [
            Tree('RB', ['away'])
        ])
    ])
])

print("Tree Output:")
print(manual_tree)

print("\n\nPretty Print:")
manual_tree.pretty_print()
```

2) Step 2 — Preprocess (keep <s> and </s>)
What preprocessing we do (simple + safe)

We:

split by lines (each line is one sentence)

split each sentence by spaces (so <s> and </s> stay as tokens)

remove any empty pieces

```python
# 2) Split into lines (sentences)
lines = [line.strip() for line in text.splitlines() if line.strip()]

# 3) Split each line into tokens by spaces (keeps <s> and </s>)
sentences = [line.split() for line in lines]

print(sentences)
```

```python
from collections import Counter

# Flatten all tokens into one list
all_tokens = [tok for sent in sentences for tok in sent]

# Count unigrams (single tokens)
unigram_counts = Counter(all_tokens)

# Count bigrams (pairs)
bigram_counts = Counter()
for sent in sentences:
    for i in range(len(sent) - 1):
        bigram_counts[(sent[i], sent[i+1])] += 1

print("count(<s>) =", unigram_counts["<s>"])
print("count(<s>, I) =", bigram_counts[("<s>", "I")])
```

```python
def sentence_probability(tokens, unigram_counts, bigram_counts, V, smoothing=False):
    prob = 1.0

    for i in range(1, len(tokens)):
        prev_word = tokens[i-1]
        word = tokens[i]

        bigram = (prev_word, word)

        if smoothing:
            numerator = bigram_counts[bigram] + 1
            denominator = unigram_counts[prev_word] + V
        else:
            numerator = bigram_counts[bigram]
            denominator = unigram_counts[prev_word]

            # If a bigram was never seen, probability becomes 0
            if numerator == 0:
                return 0.0

        prob *= numerator / denominator

    return prob
```

```python
test_sentence = "<s> I read a different book by Danielle </s>".split()

V = len(set(all_tokens))  # vocabulary size

unsmoothed = sentence_probability(test_sentence, unigram_counts, bigram_counts, V, smoothing=False)
smoothed = sentence_probability(test_sentence, unigram_counts, bigram_counts, V, smoothing=True)

print("Vocabulary size V =", V)
print("Unsmoothed probability =", unsmoothed)
print("Smoothed probability   =", smoothed)
```
